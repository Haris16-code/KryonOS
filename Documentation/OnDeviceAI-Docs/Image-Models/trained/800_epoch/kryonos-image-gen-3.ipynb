```python
import os
import sys
import time
import struct
import subprocess
import shutil
import json
import io
import re

# =============================================================================
# KAGGLE SECRETS & HUGGING FACE TOKEN RESOLUTION
# =============================================================================
# Optional fallback: If you don't want to use Kaggle Secrets, you can paste your
# Hugging Face Write Token ("hf_...") directly inside the quotes below:
HF_TOKEN_OVERRIDE = ""

# Automatically read `secret_value_0` if you pasted Kaggle's snippet above this code,
# or fetch `HF_TOKEN` directly via Kaggle's UserSecretsClient:
secret_value_0 = globals().get("secret_value_0", "")
if not secret_value_0:
    try:
        from kaggle_secrets import UserSecretsClient
        user_secrets = UserSecretsClient()
        secret_value_0 = user_secrets.get_secret("HF_TOKEN")
    except Exception as _kaggle_secret_err:
        print(f"[*] Kaggle Secret 'HF_TOKEN' note: {_kaggle_secret_err}")
        print("    -> Tip: In Kaggle, open 'Add-ons' -> 'Secrets' and make sure the CHECKBOX")
        print("       next to 'HF_TOKEN' is ticked (attached to this notebook), OR paste your")
        print("       token directly into HF_TOKEN_OVERRIDE at the top of this script.")

# =============================================================================
# CONFIGURATION
# =============================================================================
DATASET_HANDLE = "yehongjiang/pokemon-sprites-images"
OUTPUT_DIR = "/kaggle/working" if os.path.exists("/kaggle/working") else "."
SCRIPT_PATH = os.path.join(OUTPUT_DIR, "train_sprites_worker.py")
OUTPUT_KRAI = os.path.join(OUTPUT_DIR, "kryonimage_3m_96.krai")

# Hugging Face Private Model Repository Name
# Automatically becomes "<your-hf-username>/KryonOS-Image-Gen-AI-Model"
HF_REPO_NAME = os.environ.get("HF_REPO_ID", "KryonOS-Image-Gen-AI-Model")

# Training Hyperparameters
EPOCHS = 800
RESUME_FROM = "auto"   # "auto" = latest from HF Private Repo / local | "none" | "/path/to/file.pt"


# =============================================================================
# DATASET RESOLVER
# =============================================================================
def resolve_dataset():
    print(f"[1/4] Resolving dataset: {DATASET_HANDLE}...")
    for p in [
        "/kaggle/input/datasets/yehongjiang/pokemon-sprites-images",
        "/kaggle/input/pokemon-sprites-images",
        "/kaggle/input/yehongjiang-pokemon-sprites-images",
        "/kaggle/input/yehongjiang/pokemon-sprites-images",
    ]:
        if os.path.exists(p):
            for r, _, files in os.walk(p):
                if any(f.lower().endswith(".png") for f in files):
                    print(f"[+] Found dataset at: {p}")
                    return p

    print(f"[*] Downloading {DATASET_HANDLE} via kagglehub...")
    try:
        import kagglehub
    except ImportError:
        subprocess.check_call([
            sys.executable, "-m", "pip", "install", "-q",
            "kagglehub", "pillow", "tqdm", "huggingface_hub"
        ])
        import kagglehub

    path = kagglehub.dataset_download(DATASET_HANDLE)
    print(f"[+] Downloaded to: {path}")
    return path


# =============================================================================
# HUGGING FACE PRIVATE STORAGE MANAGER
# =============================================================================
class HuggingFaceStorageManager:
    """Manages private model storage on Hugging Face Hub using an HF Write Token."""
    def __init__(self, api, token, repo_id):
        self.api = api
        self.token = token
        self.repo_id = repo_id

    @classmethod
    def from_secrets_or_env(cls, default_repo_name):
        token = ""
        found_key = None

        # 1. Check HF_TOKEN_OVERRIDE or secret_value_0 first
        if HF_TOKEN_OVERRIDE and HF_TOKEN_OVERRIDE.strip():
            token = HF_TOKEN_OVERRIDE.strip()
            found_key = "HF_TOKEN_OVERRIDE"
        elif secret_value_0 and str(secret_value_0).strip():
            token = str(secret_value_0).strip()
            found_key = "Kaggle Secret 'HF_TOKEN' (secret_value_0)"

        # 2. Check alternative Kaggle Secret names if not found yet
        repo_name = default_repo_name
        try:
            from kaggle_secrets import UserSecretsClient
            user_secrets = UserSecretsClient()

            for rkey in ["HF_REPO_ID", "HUGGINGFACE_REPO_ID"]:
                try:
                    rval = user_secrets.get_secret(rkey)
                    if rval and rval.strip():
                        repo_name = rval.strip()
                        break
                except Exception:
                    continue

            if not token:
                for tkey in [
                    "HF_TOKEN",
                    "HUGGINGFACE_TOKEN",
                    "HUGGING_FACE_HUB_TOKEN",
                    "HF_API_TOKEN",
                    "hf_token",
                ]:
                    try:
                        val = user_secrets.get_secret(tkey)
                        if val and val.strip():
                            token = val.strip()
                            found_key = f"Kaggle Secret '{tkey}'"
                            break
                    except Exception:
                        continue
        except Exception:
            pass

        # 3. Check environment variables if not found in Kaggle Secrets
        if not token:
            for env_key in ["HF_TOKEN", "HUGGINGFACE_TOKEN", "HUGGING_FACE_HUB_TOKEN", "HF_API_TOKEN"]:
                val = os.environ.get(env_key, "").strip()
                if val:
                    token = val
                    found_key = f"ENV:{env_key}"
                    break

        if not token:
            return None, (
                "No Hugging Face token found. Make sure 'HF_TOKEN' is checked in "
                "Kaggle Add-ons -> Secrets, or paste it into HF_TOKEN_OVERRIDE."
            )

        # Ensure huggingface_hub is installed
        try:
            import huggingface_hub  # noqa: F401
        except ImportError:
            subprocess.check_call([
                sys.executable, "-m", "pip", "install", "-q", "huggingface_hub"
            ])

        from huggingface_hub import HfApi
        try:
            api = HfApi(token=token)
            user_info = api.whoami(token=token)
            username = user_info.get("name")
            if not username:
                return None, "Could not resolve username from Hugging Face token."

            if "/" not in repo_name:
                repo_id = f"{username}/{repo_name}"
            else:
                repo_id = repo_name

            # Create or verify the PRIVATE repository on Hugging Face Hub
            api.create_repo(
                repo_id=repo_id,
                token=token,
                private=True,
                repo_type="model",
                exist_ok=True
            )
            print(f"[+] Authenticated via {found_key} as '{username}'")
            print(f"[+] Hugging Face Private Repo Ready -> https://huggingface.co/{repo_id}")
            return cls(api, token, repo_id), None
        except Exception as e:
            return None, f"Hugging Face initialization failed: {e}"

    def verify_write_access(self):
        """Verifies write access to the private Hugging Face repo before starting training."""
        try:
            readme_content = (
                f"---\nprivate: true\ntags:\n- gan\n- pokemon\n- esp32\n- kryonos\n---\n"
                f"# {self.repo_id}\n\n"
                f"Private storage for **KryonOS 96x96 Pokemon Sprite Generator** (`kryonimage_3m_96.krai`).\n"
            ).encode("utf-8")
            self.api.upload_file(
                path_or_fileobj=io.BytesIO(readme_content),
                path_in_repo="README.md",
                repo_id=self.repo_id,
                repo_type="model",
                commit_message="Initialize KryonOS private training repository"
            )
            return True, None
        except Exception as e:
            return False, str(e)

    def download_latest_checkpoint(self, local_dir):
        """Finds and downloads the highest-epoch checkpoint from the private HF repository."""
        from huggingface_hub import hf_hub_download
        try:
            files = self.api.list_repo_files(repo_id=self.repo_id, repo_type="model")
            ckpt_files = [
                f for f in files
                if f.startswith("checkpoints/checkpoint_epoch_") and f.endswith(".pt")
            ]
            if not ckpt_files:
                return None

            ckpt_files.sort()
            latest_remote = ckpt_files[-1]
            fname = os.path.basename(latest_remote)
            local_target = os.path.join(local_dir, fname)

            if os.path.exists(local_target):
                print(f"[+] Found cached checkpoint locally: {local_target}")
                return local_target

            print(f"[+] Downloading latest checkpoint from HF Private Repo: {latest_remote}...")
            downloaded_path = hf_hub_download(
                repo_id=self.repo_id,
                filename=latest_remote,
                repo_type="model",
                token=self.token,
                local_dir=local_dir
            )
            if os.path.abspath(downloaded_path) != os.path.abspath(local_target):
                shutil.copy2(downloaded_path, local_target)
            print(f"[+] Checkpoint downloaded to: {local_target}")
            return local_target
        except Exception as e:
            print(f"[!] Could not fetch checkpoint from HF repo: {e}")
            return None


# =============================================================================
# LAUNCHER (Runs once in the main Notebook / Parent Process)
# =============================================================================
if "LOCAL_RANK" not in os.environ and "RANK" not in os.environ:
    print("=" * 74)
    print("  KRYONOS ULTRA-SHARP 96x96 POKEMON SPRITE GENERATOR (MAX QUALITY EDITION)")
    print(f"  Dataset : {DATASET_HANDLE} (Balanced 18-Type Color/Dex Conditioning)")
    print("  Engine  : ResNet-SN Projection D + DiffAugment + EMA G")
    print(f"  Storage : Hugging Face Private Hub -> {HF_REPO_NAME}")
    print("=" * 74)

    # ── 1. Connect Hugging Face Private Storage ──
    hf_mgr, hf_err = HuggingFaceStorageManager.from_secrets_or_env(HF_REPO_NAME)
    hf_ok = False
    hf_repo_id = ""
    hf_token = ""

    if hf_mgr:
        write_ok, write_err = hf_mgr.verify_write_access()
        if write_ok:
            hf_ok = True
            hf_repo_id = hf_mgr.repo_id
            hf_token = hf_mgr.token
            print(f"[+] HF Private Storage Verified -> {hf_repo_id}")
        else:
            print(f"[!] HF Write Permission Check Failed: {write_err}")
            print("    -> Make sure your HF_TOKEN has 'Write' permissions enabled on huggingface.co/settings/tokens")
            print(f"    -> Continuing seamlessly with local storage in: {OUTPUT_DIR}")
    else:
        print(f"[!] Hugging Face Hub skipped ({hf_err}). Saving locally to {OUTPUT_DIR}")

    # ── 2. Resolve dataset ──
    dataset_root = resolve_dataset()

    # ── 3. Find resume checkpoint (HF Private Repo first, then local fallback) ──
    resume_ckpt = ""
    if RESUME_FROM == "auto":
        if hf_mgr and hf_ok:
            resume_ckpt = hf_mgr.download_latest_checkpoint(OUTPUT_DIR) or ""
            if resume_ckpt:
                print(f"[+] Will resume from HF Private Repo checkpoint: {resume_ckpt}")

        if not resume_ckpt:
            local_ckpts = sorted([
                os.path.join(OUTPUT_DIR, f) for f in os.listdir(OUTPUT_DIR)
                if f.startswith("checkpoint_epoch_") and f.endswith(".pt")
            ])
            if local_ckpts:
                resume_ckpt = local_ckpts[-1]
                print(f"[+] Will resume from local checkpoint: {resume_ckpt}")

    elif RESUME_FROM not in ("none", ""):
        resume_ckpt = RESUME_FROM

    if not resume_ckpt:
        print("[*] Starting fresh training with full 18-class balanced conditioning")

    # ── 4. Environment variables for DDP workers ──
    os.environ["DATASET_ROOT_PATH"] = dataset_root
    os.environ["HF_ACTIVE_REPO_ID"] = hf_repo_id if hf_ok else ""
    os.environ["HF_ACTIVE_TOKEN"] = hf_token if hf_ok else ""
    os.environ["RESUME_CHECKPOINT"] = resume_ckpt if resume_ckpt else ""
    os.environ["TRAIN_EPOCHS"] = str(EPOCHS)
    os.environ["OMP_NUM_THREADS"] = "2"
    os.environ["PYTHONUNBUFFERED"] = "1"

    # ── 5. Detect GPU count ──
    try:
        import torch
        num_gpus = max(1, torch.cuda.device_count())
    except Exception:
        num_gpus = 1

    # ── 6. Write worker script ──
    print("[2/4] Writing stabilized high-definition worker script...")

    worker_code = '''import os
import sys
import time
import struct
import gc
import json
import copy
import queue
import threading
import random
import re
from concurrent.futures import ThreadPoolExecutor
import numpy as np
from PIL import Image

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import torch.distributed as dist
from torchvision.utils import make_grid

# =============================================================================
# WORKER CONFIGURATION
# =============================================================================
DATASET_ROOT = os.environ.get("DATASET_ROOT_PATH", "/kaggle/input/pokemon-sprites-images")
HF_REPO_ID = os.environ.get("HF_ACTIVE_REPO_ID", "")
HF_TOKEN = os.environ.get("HF_ACTIVE_TOKEN", "")
RESUME_CKPT = os.environ.get("RESUME_CHECKPOINT", "")
OUTPUT_DIR = "/kaggle/working" if os.path.exists("/kaggle/working") else "."
OUTPUT_KRAI = os.path.join(OUTPUT_DIR, "kryonimage_3m_96.krai")

LATENT_DIM = 64
EMBED_DIM = 64
NUM_CLASSES = 18
IMG_SIZE = 96
BATCH_SIZE = 64       # Per-GPU (64 on 1x GPU or 128 Global on 2x T4)
EPOCHS = int(os.environ.get("TRAIN_EPOCHS", "800"))

# TTUR-balanced learning rates for Spectral Norm ResNet Projection GAN
LR_G = 0.00020
LR_D = 0.00020
BETA1 = 0.0
BETA2 = 0.99
EMA_DECAY = 0.998     # Exponential Moving Average for crisp, artifact-free inference & KRAI export

CKPT_INTERVAL = 100   # Save checkpoint every 100 epochs
PREVIEW_INTERVAL = 20 # Generate and show preview every 20 epochs

CLASSES_18 = [
    "Normal", "Fire", "Water", "Grass", "Electric", "Ice",
    "Fighting", "Poison", "Ground", "Flying", "Psychic", "Bug",
    "Rock", "Ghost", "Dragon", "Steel", "Fairy", "Dark"
]

REJECT_KEYWORDS = [
    "back", "icon", "overworld", "shadow", "footprint",
    "berry", "berries", "item", "badge", "egg", "substitute"
]


# =============================================================================
# ASYNC NON-BLOCKING HUGGING FACE PRIVATE UPLOADER (Rank 0 Background Thread)
# =============================================================================
class AsyncHFUploader:
    """Uploads previews, checkpoints, and final .krai model to a private HF repo without stalling GPUs."""
    def __init__(self, repo_id, token):
        self.repo_id = repo_id
        self.token = token
        self.q = queue.Queue()
        self.disabled = not (repo_id and token)
        self.api = None
        self.thread = None
        if not self.disabled:
            self.thread = threading.Thread(target=self._worker, daemon=True)
            self.thread.start()

    def _init_api(self):
        from huggingface_hub import HfApi
        return HfApi(token=self.token)

    def _worker(self):
        while True:
            item = self.q.get()
            if item is None:
                self.q.task_done()
                break
            local_path, path_in_repo, commit_msg = item
            if not self.disabled and os.path.exists(local_path):
                for attempt in range(3):
                    try:
                        if self.api is None:
                            self.api = self._init_api()
                        self.api.upload_file(
                            path_or_fileobj=local_path,
                            path_in_repo=path_in_repo,
                            repo_id=self.repo_id,
                            repo_type="model",
                            commit_message=commit_msg
                        )
                        break
                    except Exception as e:
                        if attempt == 2:
                            print(f"  [!] HF upload failed for {path_in_repo}: {e}", flush=True)
                        else:
                            time.sleep(2.0 * (attempt + 1))
            self.q.task_done()

    def enqueue(self, local_path, path_in_repo, commit_msg="Update training artifact"):
        if not self.disabled and local_path:
            self.q.put((local_path, path_in_repo, commit_msg))

    def flush_and_close(self):
        if self.thread is not None:
            self.q.put(None)
            self.q.join()


# =============================================================================
# DIFFERENTIABLE AUGMENTATION (DiffAugment — Prevents Discriminator Memorization!)
# =============================================================================
def diff_augment(x):
    """
    Applies differentiable Color, Translation (with white background padding),
    and Cutout augmentations to both real and fake images in [-1, 1].
    Crucial for sprite datasets (~1k-4k images) so D cannot memorize exact pixels.
    """
    B, C, H, W = x.shape
    device = x.device
    dtype = x.dtype

    # 1. Random Brightness, Saturation, and Contrast (keeps white backgrounds clean)
    if torch.rand(1, device=device).item() < 0.8:
        bright = (torch.rand(B, 1, 1, 1, device=device, dtype=dtype) - 0.5) * 0.20
        x = x + bright
        mean_c = x.mean(dim=1, keepdim=True)
        sat = torch.rand(B, 1, 1, 1, device=device, dtype=dtype) * 0.4 + 0.8
        x = (x - mean_c) * sat + mean_c
        mean_all = x.mean(dim=[1, 2, 3], keepdim=True)
        contrast = torch.rand(B, 1, 1, 1, device=device, dtype=dtype) * 0.3 + 0.85
        x = (x - mean_all) * contrast + mean_all
        x = x.clamp(-1.0, 1.0)

    # 2. Random Translation up to +/- 6 pixels (padded with +1.0 pure white background)
    if torch.rand(1, device=device).item() < 0.8:
        max_shift = 6
        padded = F.pad(x, (max_shift, max_shift, max_shift, max_shift), mode="constant", value=1.0)
        shift_y = torch.randint(0, 2 * max_shift + 1, (1,), device=device).item()
        shift_x = torch.randint(0, 2 * max_shift + 1, (1,), device=device).item()
        x = padded[:, :, shift_y : shift_y + H, shift_x : shift_x + W]

    # 3. Random Small Cutout filled with white (+1.0) to force holistic anatomy learning
    if torch.rand(1, device=device).item() < 0.5:
        cut_h, cut_w = H // 6, W // 6
        cy = torch.randint(0, H - cut_h + 1, (1,), device=device).item()
        cx = torch.randint(0, W - cut_w + 1, (1,), device=device).item()
        mask = torch.ones_like(x)
        mask[:, :, cy : cy + cut_h, cx : cx + cut_w] = 0.0
        x = x * mask + (1.0 - mask) * 1.0

    return x.contiguous(memory_format=torch.channels_last)


# =============================================================================
# Generator (3.05M params — 100% compatible with ESP32-S3 N16R8 .krai format)
# =============================================================================
class Generator96(nn.Module):
    def __init__(self, latent_dim=64, num_classes=18, embed_dim=64):
        super().__init__()
        self.class_embed = nn.Embedding(num_classes, embed_dim)
        in_dim = latent_dim + embed_dim
        self.init_proj = nn.Sequential(
            nn.Linear(in_dim, 6 * 6 * 512, bias=False),
            nn.BatchNorm1d(6 * 6 * 512),
            nn.ReLU(inplace=True)
        )
        self.up1 = nn.Sequential(
            nn.Upsample(scale_factor=2, mode="nearest"),
            nn.Conv2d(512, 256, 3, 1, 1, bias=False),
            nn.BatchNorm2d(256), nn.ReLU(inplace=True)
        )
        self.up2 = nn.Sequential(
            nn.Upsample(scale_factor=2, mode="nearest"),
            nn.Conv2d(256, 128, 3, 1, 1, bias=False),
            nn.BatchNorm2d(128), nn.ReLU(inplace=True)
        )
        self.up3 = nn.Sequential(
            nn.Upsample(scale_factor=2, mode="nearest"),
            nn.Conv2d(128, 64, 3, 1, 1, bias=False),
            nn.BatchNorm2d(64), nn.ReLU(inplace=True)
        )
        self.up4 = nn.Sequential(
            nn.Upsample(scale_factor=2, mode="nearest"),
            nn.Conv2d(64, 32, 3, 1, 1, bias=False),
            nn.BatchNorm2d(32), nn.ReLU(inplace=True)
        )
        self.out_conv = nn.Sequential(
            nn.Conv2d(32, 3, 3, 1, 1), nn.Tanh()
        )
        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, (nn.Conv2d, nn.Linear)):
                nn.init.orthogonal_(m.weight, gain=0.8)
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0.0)
            elif isinstance(m, nn.Embedding):
                nn.init.normal_(m.weight, mean=0.0, std=0.2)

    def forward(self, z, labels):
        c = self.class_embed(labels)
        x = torch.cat([z, c], dim=1)
        x = self.init_proj(x).view(-1, 512, 6, 6)
        x = self.up1(x)
        x = self.up2(x)
        x = self.up3(x)
        x = self.up4(x)
        return self.out_conv(x)


# =============================================================================
# ResNet Spectral-Norm Projection Discriminator + Minibatch StdDev
# =============================================================================
def sn_conv2d(ic, oc, k=3, s=1, p=1):
    return nn.utils.spectral_norm(nn.Conv2d(ic, oc, k, s, p))


class DiscResBlock(nn.Module):
    """Residual downsampling block with Spectral Normalization for sharp sprite gradients."""
    def __init__(self, in_ch, out_ch, downsample=True):
        super().__init__()
        self.downsample = downsample
        self.conv1 = sn_conv2d(in_ch, out_ch, 3, 1, 1)
        self.conv2 = sn_conv2d(out_ch, out_ch, 3, 1, 1)
        self.skip = sn_conv2d(in_ch, out_ch, 1, 1, 0) if (in_ch != out_ch or downsample) else None
        self.act = nn.LeakyReLU(0.2, inplace=True)

    def forward(self, x):
        h = self.act(x)
        h = self.conv1(h)
        h = self.act(h)
        h = self.conv2(h)
        if self.downsample:
            h = F.avg_pool2d(h, 2)
        s = x
        if self.skip is not None:
            s = self.skip(s)
            if self.downsample:
                s = F.avg_pool2d(s, 2)
        return h + s


class MinibatchStdDev(nn.Module):
    """Appends across-batch feature standard deviation channel to eliminate Mode Collapse."""
    def __init__(self, group_size=8):
        super().__init__()
        self.group_size = group_size

    def forward(self, x):
        B, C, H, W = x.shape
        G = min(self.group_size, B)
        if B % G != 0:
            G = B
        y = x.float().view(G, -1, C, H, W)
        y = y - y.mean(dim=0, keepdim=True)
        y = torch.sqrt(y.square().mean(dim=0) + 1e-8)
        y = y.mean(dim=[1, 2, 3], keepdim=True)
        y = y.to(x.dtype).repeat(G, 1, H, W)
        return torch.cat([x, y], dim=1)


class ProjectionDiscriminator96(nn.Module):
    def __init__(self, num_classes=18):
        super().__init__()
        self.from_rgb = nn.Sequential(
            sn_conv2d(3, 64, 3, 1, 1),
            nn.LeakyReLU(0.2, inplace=True),
            sn_conv2d(64, 64, 3, 1, 1),
            nn.AvgPool2d(2)                    # 96 -> 48
        )
        self.block1 = DiscResBlock(64, 128, downsample=True)    # 48 -> 24
        self.block2 = DiscResBlock(128, 256, downsample=True)   # 24 -> 12
        self.block3 = DiscResBlock(256, 512, downsample=True)   # 12 -> 6
        self.mb_std = MinibatchStdDev(group_size=8)
        self.final_conv = nn.Sequential(
            sn_conv2d(513, 512, 3, 1, 1),
            nn.LeakyReLU(0.2, inplace=True)
        )
        self.linear = nn.utils.spectral_norm(nn.Linear(512, 1))
        self.embed = nn.utils.spectral_norm(nn.Embedding(num_classes, 512))

    def forward(self, img, labels):
        h = self.from_rgb(img)
        h = self.block1(h)
        h = self.block2(h)
        h = self.block3(h)
        h = self.mb_std(h)
        h = self.final_conv(h)
        h = h.sum(dim=[2, 3])                  # Global sum pooling (Miyato et al.)
        out = self.linear(h)
        proj = (h * self.embed(labels)).sum(dim=1, keepdim=True)
        return (out + proj).squeeze(1)


# =============================================================================
# 18-CLASS COLOR + DEX SEMANTIC CLASSIFIER (Fixes the 16-Empty-Class Bug!)
# =============================================================================
def classify_sprite_to_18_types(rgb_arr_chw, fpath):
    """
    Assigns every sprite to one of the 18 Pokemon types (0..17) based on its foreground
    HSV color profile + filename seed so that all 18 classes receive rich, semantically
    coherent training examples and zero classes are ever left empty.
    """
    r = rgb_arr_chw[0].astype(np.float32) / 255.0
    g = rgb_arr_chw[1].astype(np.float32) / 255.0
    b = rgb_arr_chw[2].astype(np.float32) / 255.0

    # Mask out near-white background pixels
    fg_mask = (r < 0.94) | (g < 0.94) | (b < 0.94)
    if np.sum(fg_mask) < 30:
        fg_mask = np.ones_like(r, dtype=bool)

    rf, gf, bf = r[fg_mask], g[fg_mask], b[fg_mask]
    max_c = np.maximum(np.maximum(rf, gf), bf)
    min_c = np.minimum(np.minimum(rf, gf), bf)
    delta = max_c - min_c

    val = float(np.mean(max_c))
    sat = float(np.mean(np.where(max_c > 1e-4, delta / (max_c + 1e-4), 0.0)))

    # Compute circular mean Hue in degrees [0, 360)
    colorful = delta > 0.12
    if np.sum(colorful) > 15:
        rc = rf[colorful]
        gc = gf[colorful]
        bc = bf[colorful]
        dc = delta[colorful]
        mc = max_c[colorful]
        hue = np.zeros_like(dc)
        idx_r = (mc == rc)
        idx_g = (mc == gc) & (~idx_r)
        idx_b = (~idx_r) & (~idx_g)
        hue[idx_r] = (60.0 * ((gc[idx_r] - bc[idx_r]) / (dc[idx_r] + 1e-6)) + 360.0) % 360.0
        hue[idx_g] = (60.0 * ((bc[idx_g] - rc[idx_g]) / (dc[idx_g] + 1e-6)) + 120.0) % 360.0
        hue[idx_b] = (60.0 * ((rc[idx_b] - gc[idx_b]) / (dc[idx_b] + 1e-6)) + 240.0) % 360.0
        rad = np.deg2rad(hue)
        mean_hue = float((np.rad2deg(np.arctan2(np.mean(np.sin(rad)), np.mean(np.cos(rad)))) + 360.0) % 360.0)
    else:
        mean_hue = -1.0

    # Deterministic tie-breaker from filename digits/letters
    fname = os.path.basename(fpath).lower()
    nums = re.findall(r"\\d+", fname)
    dex_seed = int(nums[0]) if nums else sum(ord(ch) for ch in fname)

    # Low saturation -> Normal (0), Rock (12), Steel (15), Dark (17)
    if sat < 0.22 or mean_hue < 0:
        if val < 0.38:
            return 17  # Dark
        elif val < 0.58:
            return 12 if (dex_seed % 2 == 0) else 15  # Rock / Steel
        else:
            return 0 if (dex_seed % 2 == 0) else 15   # Normal / Steel

    # Chromatic mapping across all 18 classes
    if mean_hue < 22 or mean_hue >= 345:
        return 1 if sat > 0.42 else 6                 # Fire (1) / Fighting (6)
    elif 22 <= mean_hue < 45:
        return 8 if val < 0.62 else 1                 # Ground (8) / Fire (1)
    elif 45 <= mean_hue < 72:
        return 4 if sat > 0.45 else 12                # Electric (4) / Rock (12)
    elif 72 <= mean_hue < 155:
        return 3 if (dex_seed % 2 == 0) else 11       # Grass (3) / Bug (11)
    elif 155 <= mean_hue < 200:
        return 5 if val > 0.62 else 2                 # Ice (5) / Water (2)
    elif 200 <= mean_hue < 250:
        return 2 if (dex_seed % 2 == 0) else 9        # Water (2) / Flying (9)
    elif 250 <= mean_hue < 295:
        if val < 0.48:
            return 13                                 # Ghost (13)
        return 14 if (dex_seed % 2 == 0) else 7       # Dragon (14) / Poison (7)
    else:
        return 10 if (dex_seed % 2 == 0) else 16      # Psychic (10) / Fairy (16)


def _decode_single_sprite(fpath):
    """Auto-crops transparent or solid-background sprites, centers on pure white, and resizes to 96x96."""
    try:
        with Image.open(fpath) as im:
            rgba = im.convert("RGBA")
            arr_rgba = np.array(rgba)
            alpha = arr_rgba[:, :, 3]

            # Detect foreground via alpha or non-white/non-black border difference
            if np.min(alpha) < 250:
                fg = alpha > 20
            else:
                rgb_tmp = arr_rgba[:, :, :3].astype(np.int16)
                bg_color = rgb_tmp[0, 0]
                diff = np.max(np.abs(rgb_tmp - bg_color), axis=2)
                fg = diff > 15
                arr_rgba[~fg, 3] = 0
                rgba = Image.fromarray(arr_rgba, mode="RGBA")

            rows = np.any(fg, axis=1)
            cols = np.any(fg, axis=0)

            if not (np.any(rows) and np.any(cols)):
                return None

            ymin, ymax = np.where(rows)[0][[0, -1]]
            xmin, xmax = np.where(cols)[0][[0, -1]]
            w = xmax - xmin + 1
            h = ymax - ymin + 1
            if w < 14 or h < 14:
                return None

            margin = max(2, int(max(w, h) * 0.07))
            max_dim = max(w, h) + 2 * margin

            square = Image.new("RGBA", (max_dim, max_dim), (255, 255, 255, 0))
            paste_x = (max_dim - w) // 2
            paste_y = (max_dim - h) // 2
            cropped = rgba.crop((xmin, ymin, xmax + 1, ymax + 1))
            square.paste(cropped, (paste_x, paste_y))

            canvas = Image.new("RGBA", (max_dim, max_dim), (255, 255, 255, 255))
            final_img = Image.alpha_composite(canvas, square).convert("RGB")

            rgb = final_img.resize((96, 96), Image.Resampling.LANCZOS)
            arr = np.array(rgb, dtype=np.uint8).transpose(2, 0, 1)  # CHW

            label = classify_sprite_to_18_types(arr, fpath)
            return arr, arr[:, :, ::-1].copy(), label
    except Exception:
        return None


def load_filtered_sprites_sharded(root_dir, rank, world_size):
    """Loads all front-facing sprites, balances all 18 classes, and shards across GPUs."""
    candidate_files = []
    for r, _, files in os.walk(root_dir):
        for f in sorted(files):
            if not f.lower().endswith((".png", ".jpg", ".jpeg")):
                continue
            full = os.path.join(r, f)
            rel = full.lower().replace("\\\\", "/")
            if any(kw in rel for kw in REJECT_KEYWORDS):
                continue
            candidate_files.append(full)

    front_only = [f for f in candidate_files if "front" in f.lower().replace("\\\\", "/")]
    selected_files = front_only if len(front_only) >= 800 else candidate_files
    selected_files.sort()

    shard_files = selected_files[rank::world_size]
    if rank == 0:
        print(f"  [*] Scanning {len(selected_files):,} sprite files ({len(shard_files):,} per GPU shard)...", flush=True)

    imgs = []
    labels = []

    with ThreadPoolExecutor(max_workers=8) as ex:
        for res in ex.map(_decode_single_sprite, shard_files):
            if res is not None:
                arr, arr_flip, label = res
                imgs.append(arr)
                labels.append(label)
                imgs.append(arr_flip)
                labels.append(label)

    tensor_x = torch.from_numpy(np.stack(imgs))
    tensor_y = torch.tensor(labels, dtype=torch.long)

    # Guarantee every one of the 18 classes has ample samples
    counts = torch.bincount(tensor_y, minlength=NUM_CLASSES)
    for c_idx in range(NUM_CLASSES):
        if counts[c_idx] < 16:
            rescue_indices = torch.arange(c_idx, len(tensor_y), NUM_CLASSES)
            tensor_y[rescue_indices] = c_idx
    counts = torch.bincount(tensor_y, minlength=NUM_CLASSES)

    if rank == 0:
        min_c, max_c = int(counts.min().item()), int(counts.max().item())
        print(f"  [+] Loaded {len(imgs) * world_size:,} total augmented sprites into VRAM!", flush=True)
        print(f"  [+] All {NUM_CLASSES}/18 Classes Active! (Min/class: {min_c}, Max/class: {max_c} per shard)", flush=True)

    return tensor_x, tensor_y, len(imgs) * world_size


# =============================================================================
# KRAI Binary Export (INT8 quantized for ESP32-S3 deployment)
# =============================================================================
def export_krai_binary(generator, out_path):
    generator.eval().cpu()
    state = generator.state_dict()

    header_fmt = "<IHHIIHHHHHHIIIHH20s"

    vocab_bytes = bytearray()
    for idx, c_name in enumerate(CLASSES_18):
        b = c_name.encode("utf-8")
        vocab_bytes.extend(struct.pack("<HH", idx, len(b)) + b)

    vocab_offset = 64
    weights_offset = (vocab_offset + len(vocab_bytes) + 15) & ~15
    weights_bytes = bytearray()
    total_params = 0

    for name, tensor in state.items():
        if "num_batches_tracked" in name:
            continue
        arr = tensor.float().numpy()
        total_params += arr.size
        max_val = float(np.max(np.abs(arr)))
        scale = (max_val / 127.0) if max_val > 1e-8 else 1.0
        q_arr = np.clip(np.round(arr / scale), -128, 127).astype(np.int8)

        n_enc = name.encode("utf-8")
        weights_bytes.append(len(n_enc))
        weights_bytes.extend(n_enc)
        weights_bytes.extend(struct.pack("<fB", scale, arr.ndim))
        for d in arr.shape:
            weights_bytes.extend(struct.pack("<I", d))
        weights_bytes.extend(q_arr.tobytes())

        pad = (16 - (len(weights_bytes) % 16)) % 16
        weights_bytes.extend(bytes(pad))

    raw_tag = b"DCGAN-96-3M" + bytes(9)
    header = struct.pack(
        header_fmt,
        0x4B524149,
        1, 1, total_params, 1,
        LATENT_DIM, 512, 5, 0, len(CLASSES_18), 0,
        vocab_offset, weights_offset, len(weights_bytes),
        IMG_SIZE, IMG_SIZE, raw_tag
    )

    with open(out_path, "wb") as f:
        f.write(header)
        f.seek(vocab_offset)
        f.write(vocab_bytes)
        f.seek(weights_offset)
        f.write(weights_bytes)

    size_mb = os.path.getsize(out_path) / (1024 * 1024)
    print(f"[+] KRAI EXPORT (EMA Weights): {out_path} ({size_mb:.2f} MB)", flush=True)
    print(f"    Parameters : {total_params:,} (~{total_params/1e6:.2f}M)", flush=True)
    print(f"    Framebuffer: {IMG_SIZE}x{IMG_SIZE} RGB565 ({IMG_SIZE*IMG_SIZE*2:,} bytes)", flush=True)


@torch.no_grad()
def update_ema(ema_model, live_model, decay):
    """Updates Exponential Moving Average generator parameters and BatchNorm buffers."""
    ema_params = dict(ema_model.named_parameters())
    for name, param in live_model.named_parameters():
        ema_params[name].mul_(decay).add_(param.data, alpha=1.0 - decay)
    ema_buffers = dict(ema_model.named_buffers())
    for name, buf in live_model.named_buffers():
        if buf.dtype.is_floating_point:
            ema_buffers[name].mul_(decay).add_(buf.data, alpha=1.0 - decay)
        else:
            ema_buffers[name].copy_(buf.data)


# =============================================================================
# MAIN TRAINING LOOP
# =============================================================================
def main():
    dist.init_process_group("nccl")
    rank = int(os.environ["RANK"])
    world_size = int(os.environ["WORLD_SIZE"])
    local_rank = int(os.environ["LOCAL_RANK"])

    torch.cuda.set_device(local_rank)
    device = torch.device(f"cuda:{local_rank}")

    # Hardware speed optimizations
    torch.backends.cudnn.benchmark = True
    torch.backends.cuda.matmul.allow_tf32 = True
    torch.backends.cudnn.allow_tf32 = True
    if hasattr(torch, "set_float32_matmul_precision"):
        torch.set_float32_matmul_precision("high")

    uploader = AsyncHFUploader(HF_REPO_ID, HF_TOKEN) if rank == 0 else None

    # ── Load shard directly into GPU VRAM as FP16 + channels_last (NHWC) ──
    if rank == 0:
        print(f"[*] Loading & classifying sprites from: {DATASET_ROOT}", flush=True)
    raw_x, raw_y, total_len = load_filtered_sprites_sharded(DATASET_ROOT, rank, world_size)

    gpu_x = (
        (raw_x.to(device, dtype=torch.float16, non_blocking=True) / 127.5) - 1.0
    ).contiguous(memory_format=torch.channels_last)
    gpu_y = raw_y.to(device, non_blocking=True)
    del raw_x, raw_y
    torch.cuda.empty_cache()

    local_samples = len(gpu_x)
    steps_per_epoch = max(1, local_samples // BATCH_SIZE)
    vram_mb = (gpu_x.element_size() * gpu_x.nelement() + gpu_y.element_size() * gpu_y.nelement()) / (1024 * 1024)
    if rank == 0:
        print(f"[+] 100% FP16 NHWC VRAM Resident: {vram_mb:.1f} MB/GPU | Steps/Epoch: {steps_per_epoch}", flush=True)

    # ── Build models (Generator96 + EMA copy + ResNet Projection Discriminator) ──
    netG_raw = Generator96().to(device, memory_format=torch.channels_last)
    netG_ema = copy.deepcopy(netG_raw).to(device, memory_format=torch.channels_last).eval()
    for p in netG_ema.parameters():
        p.requires_grad_(False)

    netD_raw = ProjectionDiscriminator96().to(device, memory_format=torch.channels_last)

    if world_size > 1:
        netG = nn.parallel.DistributedDataParallel(
            netG_raw, device_ids=[local_rank], broadcast_buffers=True, gradient_as_bucket_view=True
        )
        netD = nn.parallel.DistributedDataParallel(
            netD_raw, device_ids=[local_rank], broadcast_buffers=False, gradient_as_bucket_view=True
        )
    else:
        netG = netG_raw
        netD = netD_raw

    try:
        optG = optim.Adam(netG.parameters(), lr=LR_G, betas=(BETA1, BETA2), fused=True)
        optD = optim.Adam(netD.parameters(), lr=LR_D, betas=(BETA1, BETA2), fused=True)
    except TypeError:
        optG = optim.Adam(netG.parameters(), lr=LR_G, betas=(BETA1, BETA2))
        optD = optim.Adam(netD.parameters(), lr=LR_D, betas=(BETA1, BETA2))

    try:
        scaler_G = torch.amp.GradScaler("cuda")
        scaler_D = torch.amp.GradScaler("cuda")
    except (AttributeError, TypeError):
        scaler_G = torch.cuda.amp.GradScaler()
        scaler_D = torch.cuda.amp.GradScaler()

    z_buf = torch.empty(BATCH_SIZE, LATENT_DIM, device=device, dtype=torch.float16)

    # Fixed latent grid: 18 distinct Pokemon classes (3 rows x 6 cols)
    torch.manual_seed(42)
    fixed_z = torch.randn(NUM_CLASSES, LATENT_DIM, device=device)
    fixed_labels = torch.arange(NUM_CLASSES, dtype=torch.long, device=device)

    start_epoch = 1

    # ── Resume from checkpoint ──
    if RESUME_CKPT and os.path.exists(RESUME_CKPT):
        if rank == 0:
            print(f"[*] Loading checkpoint: {RESUME_CKPT}", flush=True)
        try:
            ckpt = torch.load(RESUME_CKPT, map_location=device, weights_only=False)
        except TypeError:
            ckpt = torch.load(RESUME_CKPT, map_location=device)

        try:
            netG_raw.load_state_dict(ckpt["generator_state_dict"])
            if "generator_ema_state_dict" in ckpt:
                netG_ema.load_state_dict(ckpt["generator_ema_state_dict"])
            else:
                netG_ema.load_state_dict(ckpt["generator_state_dict"])
            netD_raw.load_state_dict(ckpt["discriminator_state_dict"])
            optG.load_state_dict(ckpt["optimizer_g_state_dict"])
            optD.load_state_dict(ckpt["optimizer_d_state_dict"])
            scaler_G.load_state_dict(ckpt["scaler_g_state_dict"])
            scaler_D.load_state_dict(ckpt["scaler_d_state_dict"])
            start_epoch = ckpt.get("epoch", 0) + 1
            if rank == 0:
                print(f"[+] Resumed cleanly from epoch {start_epoch - 1}", flush=True)
        except Exception as e:
            if rank == 0:
                print(f"[!] Old architecture checkpoint detected ({e}); starting fresh with upgraded ResNet-D + EMA!", flush=True)
            start_epoch = 1

    if rank == 0:
        print(f"[*] Engine: {world_size}x GPU DDP | DiffAugment + ResNet-SN D + MinibatchStd + EMA G", flush=True)
        print(f"    Samples: {total_len:,} | Per-GPU: {local_samples:,} | Global Batch: {BATCH_SIZE * world_size}", flush=True)
        print(f"    Steps/Epoch: {steps_per_epoch} | Epochs: {start_epoch} -> {EPOCHS}", flush=True)
        print(f"__EVENT_INIT__|{start_epoch}|{EPOCHS}", flush=True)

    gc.disable()
    t0 = time.time()
    global_step = 0

    try:
        for epoch in range(start_epoch, EPOCHS + 1):
            ep_start = time.time()
            perm = torch.randperm(local_samples, device=device)

            # Cosine LR decay in the final 25% of training for razor-sharp pixel convergence
            if epoch > int(EPOCHS * 0.75):
                progress = (epoch - int(EPOCHS * 0.75)) / max(1, EPOCHS - int(EPOCHS * 0.75))
                lr_factor = max(0.15, 0.5 * (1.0 + np.cos(np.pi * progress)))
                for pg in optG.param_groups:
                    pg["lr"] = LR_G * lr_factor
                for pg in optD.param_groups:
                    pg["lr"] = LR_D * lr_factor

            for step in range(steps_per_epoch):
                global_step += 1
                idx = perm[step * BATCH_SIZE : (step + 1) * BATCH_SIZE]
                real_imgs = gpu_x[idx]
                labels = gpu_y[idx]

                # ── 1. Train Discriminator (Hinge Loss + DiffAugment) ──
                optD.zero_grad(set_to_none=True)
                with torch.amp.autocast("cuda"):
                    aug_real = diff_augment(real_imgs)
                    d_out_real = netD(aug_real, labels)
                    loss_d_real = F.relu(1.0 - d_out_real).mean()

                    z_buf.normal_()
                    fake_imgs = netG(z_buf, labels)
                    aug_fake_d = diff_augment(fake_imgs.detach())
                    d_out_fake = netD(aug_fake_d, labels)
                    loss_d_fake = F.relu(1.0 + d_out_fake).mean()

                    loss_d = loss_d_real + loss_d_fake

                scaler_D.scale(loss_d).backward()
                scaler_D.step(optD)
                scaler_D.update()

                # ── 2. Train Generator (Uses unwrapped netD_raw + DiffAugment + EMA Update) ──
                optG.zero_grad(set_to_none=True)
                for p in netD_raw.parameters():
                    p.requires_grad_(False)

                with torch.amp.autocast("cuda"):
                    z_buf.normal_()
                    gen_labels = torch.randint(0, NUM_CLASSES, (BATCH_SIZE,), device=device)
                    gen_imgs = netG(z_buf, gen_labels)
                    aug_fake_g = diff_augment(gen_imgs)
                    g_out = netD_raw(aug_fake_g, gen_labels)
                    loss_g = -g_out.mean()

                scaler_G.scale(loss_g).backward()
                for p in netD_raw.parameters():
                    p.requires_grad_(True)
                scaler_G.step(optG)
                scaler_G.update()

                # Update EMA Generator (warmup decay from 0.95 -> 0.998)
                cur_decay = min(EMA_DECAY, (1.0 + global_step) / (10.0 + global_step))
                update_ema(netG_ema, netG_raw, cur_decay)

            # Sync once per epoch for throughput timing
            torch.cuda.synchronize(device)
            dt = max(time.time() - ep_start, 1e-3)
            speed = (steps_per_epoch * BATCH_SIZE * world_size) / dt

            if rank == 0:
                d_val = loss_d.item()
                g_val = loss_g.item()
                elapsed = (time.time() - t0) / 60
                print(f"__EVENT_PROGRESS__|{epoch}|{EPOCHS}|{dt:.2f}|{speed:.0f}|{d_val:.4f}|{g_val:.4f}|{elapsed:.1f}", flush=True)

            # ── Preview Every 20 Epochs (Using EMA Generator for Crisp Output!) ──
            if rank == 0 and (epoch % PREVIEW_INTERVAL == 0 or epoch == EPOCHS or epoch == 1):
                netG_ema.eval()
                with torch.no_grad():
                    with torch.amp.autocast("cuda"):
                        samples = netG_ema(fixed_z, fixed_labels)
                    samples = samples.contiguous().cpu().float()
                    samples = (samples * 0.5 + 0.5).clamp(0, 1)
                    grid = make_grid(samples, nrow=6, padding=4, pad_value=1.0)
                    grid_np = (grid.permute(1, 2, 0).numpy() * 255).astype(np.uint8)
                    grid_img = Image.fromarray(grid_np)
                    preview_path = os.path.join(OUTPUT_DIR, f"preview_epoch_{epoch:04d}.png")
                    grid_img.save(preview_path)

                print(f"__EVENT_PREVIEW__|{epoch}|{preview_path}", flush=True)

                if uploader:
                    uploader.enqueue(
                        preview_path,
                        f"previews/preview_epoch_{epoch:04d}.png",
                        f"Add preview at epoch {epoch}"
                    )

            # ── Checkpoint Every 100 Epochs ──
            if rank == 0 and (epoch % CKPT_INTERVAL == 0 or epoch == EPOCHS):
                ckpt_path = os.path.join(OUTPUT_DIR, f"checkpoint_epoch_{epoch:04d}.pt")
                torch.save({
                    "epoch": epoch,
                    "generator_state_dict": netG_raw.state_dict(),
                    "generator_ema_state_dict": netG_ema.state_dict(),
                    "discriminator_state_dict": netD_raw.state_dict(),
                    "optimizer_g_state_dict": optG.state_dict(),
                    "optimizer_d_state_dict": optD.state_dict(),
                    "scaler_g_state_dict": scaler_G.state_dict(),
                    "scaler_d_state_dict": scaler_D.state_dict(),
                    "config": {
                        "latent_dim": LATENT_DIM, "embed_dim": EMBED_DIM,
                        "num_classes": NUM_CLASSES, "img_size": IMG_SIZE,
                        "batch_size": BATCH_SIZE, "lr_g": LR_G, "lr_d": LR_D,
                    }
                }, ckpt_path)
                print(f"  [SAVE] Checkpoint saved: {ckpt_path}", flush=True)

                if uploader:
                    uploader.enqueue(
                        ckpt_path,
                        f"checkpoints/checkpoint_epoch_{epoch:04d}.pt",
                        f"Save checkpoint epoch {epoch}"
                    )

    finally:
        gc.enable()
        try:
            dist.destroy_process_group()
        except Exception:
            pass

    # ── Final Export (Exports EMA Generator to .krai & Uploads to Private HF Repo) ──
    if rank == 0:
        total_min = (time.time() - t0) / 60
        print(f"\\n[+] Training Complete in {total_min:.2f} minutes!", flush=True)

        export_krai_binary(netG_ema, OUTPUT_KRAI)

        if uploader:
            uploader.enqueue(
                OUTPUT_KRAI,
                "kryonimage_3m_96.krai",
                f"Export final INT8 KRAI model ({EPOCHS} epochs)"
            )
            print("  [*] Waiting for background Hugging Face Hub uploads to finish...", flush=True)
            uploader.flush_and_close()
            print(f"  [HF HUB] All files synced to Private HF Repo: https://huggingface.co/{HF_REPO_ID}", flush=True)


if __name__ == "__main__":
    main()
'''

    with open(SCRIPT_PATH, "w", encoding="utf-8") as f:
        f.write(worker_code)

    # ── 7. Launch torchrun & stream live progress + inline previews in Parent Process ──
    print(f"[3/4] Launching torchrun with {num_gpus} GPU(s)...")
    cmd = [
        sys.executable, "-m", "torch.distributed.run",
        f"--nproc_per_node={num_gpus}",
        "--standalone",
        SCRIPT_PATH
    ]

    from tqdm.auto import tqdm
    pbar = None

    proc = subprocess.Popen(
        cmd,
        env=os.environ.copy(),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    try:
        for raw_line in proc.stdout:
            line = raw_line.rstrip("\r\n")
            if not line:
                continue

            if line.startswith("__EVENT_INIT__|"):
                parts = line.split("|")
                start_ep, total_ep = int(parts[1]), int(parts[2])
                pbar = tqdm(
                    total=total_ep - start_ep + 1,
                    desc="Training KryonOS 96x96",
                    dynamic_ncols=True,
                    leave=True
                )

            elif line.startswith("__EVENT_PROGRESS__|"):
                parts = line.split("|")
                epoch, total_ep = int(parts[1]), int(parts[2])
                dt, speed, d_val, g_val, elapsed = parts[3], parts[4], parts[5], parts[6], parts[7]
                if pbar is not None:
                    pbar.update(1)
                    pbar.set_postfix({
                        "Epoch": f"{epoch}/{total_ep}",
                        "Speed": f"{speed} img/s",
                        "D": d_val,
                        "G": g_val
                    })
                if epoch % 10 == 0 or epoch == 1 or epoch == total_ep:
                    msg = (
                        f"Epoch [{epoch:04d}/{total_ep}] {dt}s | "
                        f"Speed: {speed} img/s | D: {d_val} | G: {g_val} | Elapsed: {elapsed}m"
                    )
                    if pbar is not None:
                        pbar.write(msg)
                    else:
                        print(msg, flush=True)

            elif line.startswith("__EVENT_PREVIEW__|"):
                parts = line.split("|")
                ep_num, preview_path = int(parts[1]), parts[2]
                msg = f"  [IMG] Preview generated (Epoch {ep_num:04d}): {preview_path}"
                if pbar is not None:
                    pbar.write(msg)
                else:
                    print(msg, flush=True)

                # Render inline in Kaggle / Jupyter Notebook output
                if os.path.exists(preview_path):
                    try:
                        from IPython.display import display, Image as IPyImage
                        display(IPyImage(filename=preview_path))
                    except Exception:
                        pass

            else:
                if pbar is not None:
                    pbar.write(line)
                else:
                    print(line, flush=True)
    finally:
        if pbar is not None:
            pbar.close()

    proc.wait()

    if proc.returncode != 0:
        print(f"[!] torchrun exited with code {proc.returncode}")
    else:
        print(f"[4/4] Complete! Output: {OUTPUT_KRAI}")
        if hf_ok:
            print(f"      Hugging Face Private Repo: https://huggingface.co/{hf_repo_id}")
```
